# Welcome to Jupyter Notebook

This notebook is a simple introduction to how notebooks work.

- A Markdown cell is for text, headings, and explanations.
- A Code cell runs Python.
- Press Shift + Enter to run a cell.

```python
print("Hello from a Jupyter code cell!")
print("Run this cell with Shift + Enter.")
```

## Variables

Variables store values that you can reuse later.

```python
name = "Ada"
age = 28
print("Name:", name)
print("Age:", age)
```

## Math and expressions

Python can do arithmetic directly.

```python
x = 10
y = 3
print("x + y =", x + y)
print("x * y =", x * y)
print("x / y =", x / y)
```

## Lists and loops

A list holds multiple values. A loop lets you work through each item.

```python
animals = ["cat", "dog", "bird"]
for animal in animals:
    print("I like", animal)
```

## Functions

Functions let you group reusable code into one block.

```python
def greet(name):
    return f"Hello, {name}!"

print(greet("Ada"))
print(greet("Linus"))
```

## Working with data

You can create a small table of numbers and analyze them.

```python
numbers = [1, 2, 3, 4, 5]
total = sum(numbers)
average = total / len(numbers)
print("Total:", total)
print("Average:", average)
```
